```python
from getpass import getpass
from qiskit_ibm_runtime import QiskitRuntimeService

API_KEY = getpass("Enter your IBM Quantum API key: ")
CRN = input("Enter your CRN: ").strip()

service = QiskitRuntimeService(
    channel="ibm_quantum_platform",
    token=API_KEY,
    instance=CRN
)

backend = service.least_busy(
    operational=True,
    simulator=False,
    min_num_qubits=1
)

print("Connected successfully!")
print("Backend:", backend.name)
```

```python
from qiskit import QuantumCircuit
import numpy as np

# Circuit A: 0 should occur more frequently
circuit_A = QuantumCircuit(1, 1)
circuit_A.ry(np.pi / 3, 0)
circuit_A.measure(0, 0)

# Circuit B: 1 should occur more frequently
circuit_B = QuantumCircuit(1, 1)
circuit_B.ry(2 * np.pi / 3, 0)
circuit_B.measure(0, 0)

print("Circuit A:")
display(circuit_A.draw("mpl"))

print("Circuit B:")
display(circuit_B.draw("mpl"))
```

```python
print("Circuit A: Ry(pi/3)")
print("Predicted P(0) = 75%")
print("Predicted P(1) = 25%")

print("\nCircuit B: Ry(2*pi/3)")
print("Predicted P(0) = 25%")
print("Predicted P(1) = 75%")
```
